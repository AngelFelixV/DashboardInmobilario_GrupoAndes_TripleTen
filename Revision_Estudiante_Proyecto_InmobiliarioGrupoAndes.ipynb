## Insights accionables

- El segmento **Primera vez** concentra la mayor parte del ingreso, por lo que el negocio depende principalmente de la captación de nuevos clientes.
- Las cohortes **2023-01**, **2023-02** y **2023-03** muestran la mayor recurrencia: los clientes adquiridos en esos meses realizaron más del doble de compras respecto a su compra inicial (**tasa de recompra > 2.2x**).
- Las ventas crecieron **11.14% YoY** en 2024 respecto a 2023.
- **Casa** genera el mayor ingreso por volumen, mientras que **Comercial** registra el ticket promedio más alto (**$1,797,097.70** vs. **$964,085.76** para Casa), representando dos motores distintos del negocio.

## Recomendaciones estratégicas

- Priorizar la comercialización de propiedades tipo **Casa**.
- Fortalecer el canal **Corredor**.
- Implementar campañas de retención para clientes **Primera vez**.
- Analizar las cohortes de inicios de 2023 para identificar factores que expliquen su mayor recurrencia.

## Nota metodológica

> Las cohortes de 2023-01, 2023-02 y 2023-03 presentan mayores tasas de recompra porque han permanecido más tiempo en la base de clientes y han tenido más oportunidades de realizar compras adicionales. Este efecto debe considerarse antes de atribuir su desempeño únicamente a factores comerciales.
